```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit
```

```python
def f(x, m, c):
    y=m*x+c
    return y
```

```python
x=np.arange(0, 10, 0.5)
m0=2.5
c0=-6

y=f(x, m0, c0)
sig=2*np.ones_like(x)
np.random.seed(1)
data=y+np.random.normal(0, scale=sig)
```

```python
plt.figure(figsize=(10,6))
plt.errorbar(x,data,fmt="ko", yerr=sig, capsize=3, label="data")
plt.plot(x, y, "r:", label="truth")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
```

## Fit using curve_fit

```python
param, param_cov=curve_fit(f, x, data, sigma=sig)
print(param)
print(param_cov)
```

```python
# what are the errors on the parameters?
```

```python
plt.figure(figsize=(10,6))
plt.errorbar(x,data,fmt="ko", yerr=sig, capsize=3, label="data")
plt.plot(x, y, "r:", label="truth")
plt.plot(x, f(x, *param), "b", label="best fit")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
```

## A new example

```python
def f(x, a, b):
    y=a*x**3+b*np.sin(x)
    return y
```

```python
x=np.arange(-2*np.pi,2*np.pi, 0.5)
a0=0.05
b0=5.4
y=f(x, a0, b0)
sig=np.sqrt(np.abs(y))*2
np.random.seed(1)
data=y+np.random.normal(0, sig)
```

```python
plt.figure(figsize=(10,6))
plt.errorbar(x,data,fmt="ko", yerr=sig, capsize=3, label="data")
plt.plot(x, y, "r:", label="truth")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
```

```python
param, param_cov=curve_fit(f, x, data, sigma=sig)
print(param)
print(param_cov)
```

```python
plt.figure(figsize=(10,6))
plt.errorbar(x,data,fmt="ko", yerr=sig, capsize=3, label="data")
plt.plot(x, y, "r:", label="truth")
plt.plot(x, f(x, *param), "b", label="best fit")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
```

```python
## see how things change in presence of off-diagonal elements in the covariance matrix.
```
